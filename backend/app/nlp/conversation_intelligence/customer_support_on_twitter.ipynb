```python
# Build a parent lookup: tweet_id -> parent tweet ID
parent_map = (
    df_twitter
    .dropna(subset=["in_response_to_tweet_id"])
    .set_index("tweet_id")["in_response_to_tweet_id"]
    .to_dict()
)

def get_depth(tweet_id, parent_map, max_depth=100):
    depth = 0
    current = tweet_id
    visited = set()

    while current in parent_map:
        if current in visited:
            break

        visited.add(current)

        parent = parent_map[current]

        if parent is None:
            break

        current = int(parent)
        depth += 1

        if depth >= max_depth:
            break

    return depth


# Analyze a manageable diagnostic subset first.
# This does NOT reduce the dataset for modelling.
sample_ids = df_twitter["tweet_id"].head(10000)

depths = sample_ids.apply(
    lambda x: get_depth(x, parent_map)
)

print("Tweets analysed:", len(depths))

print("\nDepth distribution:")
print(depths.value_counts().sort_index())

print("\nDepth statistics:")
print(depths.describe())
```

```python
def inspect_conversation(root_id, max_nodes=20):
    """
    Display the conversation chain starting from a customer root.
    """

    print("\n" + "=" * 110)
    print("CONVERSATION ROOT:", root_id)

    root_row = df.loc[df["tweet_id"] == root_id].iloc[0]

    print(
        f"[ROOT] tweet_id={root_id} | "
        f"inbound={root_row['inbound']} | "
        f"{root_row['text']}"
    )

    queue = [(int(root_id), 0)]
    visited = set()

    nodes_shown = 0

    while queue and nodes_shown < max_nodes:

        tweet_id, depth = queue.pop(0)

        if tweet_id in visited:
            continue

        visited.add(tweet_id)

        row = df.loc[df["tweet_id"] == tweet_id].iloc[0]

        print(
            f"[depth={depth}] "
            f"tweet_id={tweet_id} | "
            f"inbound={row['inbound']} | "
            f"{row['text']}"
        )

        nodes_shown += 1

        child_ids = children.get(tweet_id, [])

        for child_id in child_ids:
            queue.append((int(child_id), depth + 1))


# Inspect several customer-started conversations
for root_id in customer_roots["tweet_id"].head(10):
    inspect_conversation(int(root_id), max_nodes=20)
```

```python
def get_conversation_stats(root_id, max_depth=100):
    """
    Traverse the conversation graph and collect structural statistics.
    """

    queue = [(int(root_id), 0)]
    visited = set()

    total_nodes = 0
    customer_nodes = 0
    company_nodes = 0
    max_depth_found = 0

    while queue:

        tweet_id, depth = queue.pop(0)

        if tweet_id in visited:
            continue

        visited.add(tweet_id)

        row = df.loc[df["tweet_id"] == tweet_id].iloc[0]

        total_nodes += 1
        max_depth_found = max(max_depth_found, depth)

        if bool(row["inbound"]):
            customer_nodes += 1
        else:
            company_nodes += 1

        if depth >= max_depth:
            continue

        for child_id in children.get(tweet_id, []):
            queue.append((int(child_id), depth + 1))

    return {
        "root_id": int(root_id),
        "total_nodes": total_nodes,
        "customer_nodes": customer_nodes,
        "company_nodes": company_nodes,
        "max_depth": max_depth_found,
    }
```

```python
# ---------------------------------------------------------
# Measure DistilBERT token-length distribution
# on the training data only.
# ---------------------------------------------------------

from tqdm.auto import tqdm

train_texts = train_dataset["text"]

token_lengths = []

batch_size = 5000

for start in tqdm(
    range(0, len(train_texts), batch_size),
    desc="Measuring token lengths"
):
    batch = train_texts[start:start + batch_size]

    encoded = tokenizer(
        batch,
        truncation=False,
        padding=False,
        return_length=True,
    )

    token_lengths.extend(encoded["length"])

token_lengths = np.array(token_lengths)

print("Training examples:", len(token_lengths))

print("\nToken length statistics:")
print(
    pd.Series(token_lengths).describe(
        percentiles=[
            0.50,
            0.75,
            0.90,
            0.95,
            0.99,
            0.995,
            0.999,
        ]
    )
)

print("\nMaximum token length:", token_lengths.max())
```

```python
# ---------------------------------------------------------
# Quantify potential truncation at max_length = 128
# ---------------------------------------------------------

max_length = 128

over_limit = token_lengths > max_length

print("Chosen max_length:", max_length)

print("\nExamples requiring truncation:")
print("Count:", over_limit.sum())
print("Percentage:", round(over_limit.mean() * 100, 4))

print("\nExamples within limit:")
print("Count:", (~over_limit).sum())
print("Percentage:", round((~over_limit).mean() * 100, 4))

assert over_limit.sum() >= 0
assert (~over_limit).sum() + over_limit.sum() == len(token_lengths)
```

```python
# ---------------------------------------------------------
# Tokenize all modelling splits
# ---------------------------------------------------------

max_length = 128

def tokenize_batch(batch):
    return tokenizer(
        batch["text"],
        truncation=True,
        max_length=max_length,
        padding=False,
    )


train_tokenized = train_dataset.map(
    tokenize_batch,
    batched=True,
    batch_size=5000,
    desc="Tokenizing training data",
)

validation_tokenized = validation_dataset.map(
    tokenize_batch,
    batched=True,
    batch_size=5000,
    desc="Tokenizing validation data",
)

test_tokenized = test_dataset.map(
    tokenize_batch,
    batched=True,
    batch_size=5000,
    desc="Tokenizing test data",
)

print("Tokenization complete.")

print("\nTrain columns:")
print(train_tokenized.column_names)

print("\nValidation columns:")
print(validation_tokenized.column_names)

print("\nTest columns:")
print(test_tokenized.column_names)
```

```python
# ============================================================
# STEP 47 — CONFUSION MATRIX + CLASSIFICATION REPORT
# ============================================================

import numpy as np
import pandas as pd

from sklearn.metrics import (
    confusion_matrix,
    classification_report,
)

print("=" * 60)
print("STEP 47 — TEST SET ERROR ANALYSIS")
print("=" * 60)

print(f"Test examples: {len(test_tokenized):,}")

# ------------------------------------------------------------
# 1. Get predictions from the locked test set
# ------------------------------------------------------------

test_prediction_output = trainer.predict(test_tokenized)

test_logits = test_prediction_output.predictions
test_labels = test_prediction_output.label_ids

test_predictions = np.argmax(
    test_logits,
    axis=-1,
)

print("\nPredictions generated successfully.")

print(f"Logits shape:       {test_logits.shape}")
print(f"Labels shape:       {test_labels.shape}")
print(f"Predictions shape:  {test_predictions.shape}")


# ------------------------------------------------------------
# 2. Confusion matrix
# ------------------------------------------------------------

cm = confusion_matrix(
    test_labels,
    test_predictions,
    labels=[0, 1],
)

print("\n" + "=" * 60)
print("CONFUSION MATRIX")
print("=" * 60)

cm_df = pd.DataFrame(
    cm,
    index=["Actual SIMPLE", "Actual MULTI_TURN"],
    columns=["Predicted SIMPLE", "Predicted MULTI_TURN"],
)

print(cm_df)


# ------------------------------------------------------------
# 3. Classification report
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("CLASSIFICATION REPORT")
print("=" * 60)

report = classification_report(
    test_labels,
    test_predictions,
    labels=[0, 1],
    target_names=["SIMPLE", "MULTI_TURN"],
    digits=4,
    zero_division=0,
)

print(report)


# ------------------------------------------------------------
# 4. Explicit error counts
# ------------------------------------------------------------

true_simple = (test_labels == 0)
true_multi = (test_labels == 1)

pred_simple = (test_predictions == 0)
pred_multi = (test_predictions == 1)

true_negatives = np.sum(true_simple & pred_simple)
false_positives = np.sum(true_simple & pred_multi)
false_negatives = np.sum(true_multi & pred_simple)
true_positives = np.sum(true_multi & pred_multi)

print("=" * 60)
print("ERROR COUNTS")
print("=" * 60)

print(f"True Negatives  (SIMPLE → SIMPLE):      {true_negatives:,}")
print(f"False Positives (SIMPLE → MULTI_TURN):  {false_positives:,}")
print(f"False Negatives (MULTI_TURN → SIMPLE):  {false_negatives:,}")
print(f"True Positives  (MULTI_TURN → MULTI_TURN): {true_positives:,}")


# ------------------------------------------------------------
# 5. Sanity check
# ------------------------------------------------------------

total_predictions = (
    true_negatives
    + false_positives
    + false_negatives
    + true_positives
)

print("\n" + "=" * 60)
print("SANITY CHECK")
print("=" * 60)

print(f"Total confusion-matrix samples: {total_predictions:,}")
print(f"Expected test samples:          {len(test_tokenized):,}")

assert total_predictions == len(test_tokenized)

print("Sanity check: PASSED")
print("=" * 60)
print("STEP 47 COMPLETED")
print("=" * 60)
```

```python
# ============================================================
# STEP 49 — RELOAD SAVED MODEL + INFERENCE VERIFICATION
# ============================================================

import numpy as np
import torch

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
)

print("=" * 60)
print("STEP 49 — SAVED MODEL RELOAD VERIFICATION")
print("=" * 60)

# ------------------------------------------------------------
# 1. Define final model directory
# ------------------------------------------------------------

final_model_dir = (
    r"C:\Users\danie\supportmind-ai\backend\app\nlp"
    r"\conversation_intelligence\models"
    r"\supportmind_conversation_complexity"
)

print(f"Model directory:\n{final_model_dir}")


# ------------------------------------------------------------
# 2. Reload tokenizer independently
# ------------------------------------------------------------

reloaded_tokenizer = AutoTokenizer.from_pretrained(
    final_model_dir
)

print("\nTokenizer reloaded successfully.")
print(f"Tokenizer class: {reloaded_tokenizer.__class__.__name__}")
print(f"Vocabulary size: {reloaded_tokenizer.vocab_size}")


# ------------------------------------------------------------
# 3. Reload model independently
# ------------------------------------------------------------

reloaded_model = AutoModelForSequenceClassification.from_pretrained(
    final_model_dir
)

print("\nModel reloaded successfully.")
print(f"Model class: {reloaded_model.__class__.__name__}")
print(f"Number of labels: {reloaded_model.config.num_labels}")
print(f"Label mapping: {reloaded_model.config.id2label}")


# ------------------------------------------------------------
# 4. Move model to the available device
# ------------------------------------------------------------

device = torch.device(
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

reloaded_model.to(device)
reloaded_model.eval()

print(f"\nInference device: {device}")


# ------------------------------------------------------------
# 5. Test with real customer messages
# ------------------------------------------------------------

test_messages = [
    "I have been waiting for my order and still haven't received it.",
    "Can you please help me fix this issue?",
    "My account has been locked and I cannot log in.",
    "I just wanted to say thank you for your help.",
]


inputs = reloaded_tokenizer(
    test_messages,
    padding=True,
    truncation=True,
    max_length=128,
    return_tensors="pt",
)

inputs = {
    key: value.to(device)
    for key, value in inputs.items()
}


# ------------------------------------------------------------
# 6. Run inference
# ------------------------------------------------------------

with torch.no_grad():
    outputs = reloaded_model(**inputs)

logits = outputs.logits

probabilities = torch.softmax(
    logits,
    dim=-1,
)

predictions = torch.argmax(
    probabilities,
    dim=-1,
)


# ------------------------------------------------------------
# 7. Display predictions
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("INFERENCE RESULTS")
print("=" * 60)

for message, prediction, probability in zip(
    test_messages,
    predictions.cpu().numpy(),
    probabilities.cpu().numpy(),
):

    label = reloaded_model.config.id2label[
        int(prediction)
    ]

    confidence = float(
        probability[int(prediction)]
    )

    print(f"\nMessage:")
    print(f"  {message}")

    print(f"Prediction:")
    print(f"  {label}")

    print(f"Confidence:")
    print(f"  {confidence:.4f}")


# ------------------------------------------------------------
# 8. Verify output shape
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("OUTPUT VALIDATION")
print("=" * 60)

print(f"Logits shape:        {tuple(logits.shape)}")
print(f"Probabilities shape: {tuple(probabilities.shape)}")
print(f"Predictions shape:   {tuple(predictions.shape)}")

assert logits.shape == (len(test_messages), 2)
assert probabilities.shape == (len(test_messages), 2)
assert predictions.shape == (len(test_messages),)

print("\nOutput shape validation: PASSED")


# ------------------------------------------------------------
# 9. Verify probability sums
# ------------------------------------------------------------

probability_sums = probabilities.sum(
    dim=-1
).cpu().numpy()

print("\nProbability sums:")

for index, value in enumerate(probability_sums):
    print(f"  Message {index + 1}: {value:.6f}")

assert np.allclose(
    probability_sums,
    1.0,
    atol=1e-6,
)

print("Probability validation: PASSED")


# ------------------------------------------------------------
# Final status
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("STEP 49 COMPLETED")
print("Saved model can be independently reloaded.")
print("=" * 60)
```
